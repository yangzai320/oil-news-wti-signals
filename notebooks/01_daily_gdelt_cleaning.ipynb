```python
# %%
# 1. 读取 0–8 文件
import os
import pandas as pd
import numpy as np

cred = os.path.expanduser("~/Downloads/YOUR_PROJECT-527de85df0a1.json")

files = [f"gs://oilnews/oil_news_{i:012d}.csv" for i in range(9)]

dfs = []
for f in files:
    print("读取:", f)
    temp = pd.read_csv(f, storage_options={"token": cred})
    print("shape:", temp.shape)
    dfs.append(temp)

df = pd.concat(dfs, ignore_index=True)
print("合并后大小:", df.shape)
df.head()
```

```python
# %%
# 2. 基础清洗
df = df.dropna(subset=["title", "news_time"]).copy()

df["title"] = df["title"].astype(str).str.strip()
df = df[df["title"].str.len() >= 10]

print("基础清洗后:", df.shape)
df.head()
```

```python
# %%
# 3. 标题标准化与去重
df["title_clean"] = (
    df["title"]
    .str.lower()
    .str.replace(r"[^a-z0-9 ]", "", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

before = df.shape[0]
df = df.drop_duplicates(subset=["title_clean"]).copy()
after = df.shape[0]

print("去重前:", before)
print("去重后:", after)
print("去重掉:", before - after)
```

```python
# %%
# 4. 时间处理：新闻时间转成 UTC，再转到市场时区
import pandas as pd
import numpy as np

market_tz = "America/New_York"

df["news_time"] = pd.to_datetime(df["news_time"], errors="coerce", utc=True)
df = df.dropna(subset=["news_time"]).copy()

# 转到纽约时区，方便和收盘时点对齐
df["news_time_local"] = df["news_time"].dt.tz_convert(market_tz)

# 排序，后面 rolling window 会用到
df = df.sort_values("news_time_local").reset_index(drop=True)

print(df[["news_time", "news_time_local"]].head())
print("新闻时间范围：", df["news_time_local"].min(), "到", df["news_time_local"].max())
```

```python
# %%
# 5. 读取油价（FRED: DCOILWTICO）

oil = pd.read_csv("data/oil_price.csv")

# 重命名列（方便后面统一写法）
oil = oil.rename(columns={
    "observation_date": "date",
    "DCOILWTICO": "close"
})

# 转时间
oil["date"] = pd.to_datetime(oil["date"], errors="coerce")

# 去掉缺失（FRED 周末/节假日会是 NaN）
oil = oil.dropna(subset=["date", "close"]).copy()

oil = oil.sort_values("date").reset_index(drop=True)

print(oil.head())
print("油价数据大小:", oil.shape)
```

```python
df.to_parquet("data/oil_news_clean.parquet", index=False)
oil.to_csv("data/oil_price_clean.csv", index=False)
```

```python
# %%
# 6. 标题 embedding
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

titles = df["title"].tolist()
embeddings = model.encode(
    titles,
    batch_size=64,
    show_progress_bar=True
)

print("embedding shape:", embeddings.shape)
```

```python
# %%
# 7. embedding 放回数据表
df = df.reset_index(drop=True)
df["embedding"] = list(embeddings)

df[["title", "date", "embedding"]].head()
```
